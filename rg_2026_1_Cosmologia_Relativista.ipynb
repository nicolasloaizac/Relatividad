```python
import einsteinpy.symbolic as es
import sympy as sp
import numpy as np
```

```python
c, t, r, theta, phi ,K = sp.symbols("c t r theta phi K")

#metrica espacial 

gesp_mat = np.diag([1/(1-K*r**2), r**2, r**2*sp.sin(theta)**2])

gesp = es.MetricTensor(gesp_mat, [r, theta, phi])
gesp.tensor()
```

### Tensor de Riemman

```python
R = es.RiemannCurvatureTensor.from_metric(gesp)
R.tensor()
```

### Escalas de Riemann

```python
Resc = es.RicciScalar.from_riemann(R)
Resc.tensor()
```

### El espacio tiempo Weyl al completo 


```python
a = sp.Function("a")

#metrica espacial 

gmat = np.diag([c**2 , -a(t)**2*1/(1-K*r**2), -a(t)**2*r**2, -a(t)**2*r**2*sp.sin(theta)**2])

g = es.MetricTensor(gmat, [t,r, theta, phi])
g.tensor()
```

```python
Resc = es.RicciScalar.from_metric(g)
Resc.tensor()
```

```python
G = es.EinsteinTensor.from_metric(g)
G.tensor()
```
