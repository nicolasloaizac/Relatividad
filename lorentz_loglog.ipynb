```python
import numpy as np

import matplotlib.pyplot as plt

# beta = v/c (adimensional), evitando 0 y 1 para escalas log
beta = np.geomspace(1e-4, 1 - 1e-8, 2000)
gamma = 1.0 / np.sqrt(1.0 - beta**2)

# Constante de Planck (SI)
h = 6.62607015e-34  # J·s

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Gráfico log-log
axes[0].loglog(beta, gamma, color="tab:blue")
axes[0].set_title("Factor de Lorentz (log-log)")
axes[0].set_xlabel(r"$\beta = v/c$")
axes[0].set_ylabel(r"$\gamma = \frac{1}{\sqrt{1-\beta^2}}$")
axes[0].grid(True, which="both", ls="--", alpha=0.5)

# Gráfico semi-log (eje Y logarítmico)
axes[1].semilogy(beta, gamma, color="tab:red")
axes[1].set_title("Factor de Lorentz (semi-log en Y)")
axes[1].set_xlabel(r"$\beta = v/c$")
axes[1].set_ylabel(r"$\gamma$")
axes[1].grid(True, which="both", ls="--", alpha=0.5)

# Gráfico de la constante de Planck
axes[2].plot(beta, np.full_like(beta, h), color="tab:green")
axes[2].set_title("Constante de Planck")
axes[2].set_xlabel(r"$\beta = v/c$")
axes[2].set_ylabel(r"$h\,(\mathrm{J\cdot s})$")
axes[2].ticklabel_format(axis="y", style="sci", scilimits=(0, 0))
axes[2].grid(True, which="both", ls="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
